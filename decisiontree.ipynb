```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
%matplotlib inline
```

```python
from sklearn.datasets import load_iris
from sklearn.tree import DecisionTreeClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import classification_report, confusion_matrix
```

```python
iris = load_iris()
```

```python
print(iris['DESCR'])
```

```python
X = pd.DataFrame(iris['data'], columns=iris['feature_names']).head()
```

```python
y = pd.DataFrame(iris['target'], columns=['target']).head()
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=10)
```

```python
treeClassifier = DecisionTreeClassifier(max_depth=2)
```

```python
treeClassifier.fit(X_train, y_train)
```

```python
from sklearn import tree
plt.figure(figsize=(12,8))
tree.plot_tree(treeClassifier, filled=True)
plt.show()
```

```python
y_pred = treeClassifier.predict(X_test)
```

```python
cm = confusion_matrix(y_test, y_pred)
```

```python
print(cm)
print(classification_report(y_test, y_pred))
```

```python
param_grid = {'criterion': ['gini', 'entropy', 'log_loss'],
              'splitter': ['best', 'random'],
              'max_depth': [2, 3, 4, 5],
              'max_features': ['auto', 'sqrt', 'log2']}
```

```python
from sklearn.model_selection import GridSearchCV
treeModel = DecisionTreeClassifier()
gridSearch = GridSearchCV(estimator=treeModel, param_grid=param_grid, cv=4, scoring='accuracy')
gridSearch.fit(X_train, y_train)
```

```python
gridSearch.best_params_
```

```python
gridSearch.best_score_
```

```python
y_pred_grid = gridSearch.predict(X_test)
```

```python
cm_grid = confusion_matrix(y_test, y_pred_grid)
print(cm_grid)
print(classification_report(y_test, y_pred_grid))
```
